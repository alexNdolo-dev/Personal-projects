```python
import math
mathe = True
while mathe == True:
#1.display available operations
    print("\n=====Advanced calculator=====")
    print("Basic: +, -, *, /")
    print("Advanced: ^(power), % ")
    print("Sci: sqrt, log")

#2.Get the operator first
    op = input("Enter operator: ")
#handling binary operations(operations which require two operands(inputs))
    if op in ['+', '-', '*', '/', '^','%']:
        num1 = float(input("Enter first number:"))
        num2 = float(input("Enter second number:"))

        if op == '+':
            print(f"Result: {sum([num1, num2])}")
        elif op == '-':
            print(f"Result: {num1, num2}")
        elif op == '*':
            print(f"Result: {num1 * num2}")
        elif op == '/':
            print(f"Result: {num1 / num2}")
        elif op == '^':
            print(f"Result: {math.pow(num1, num2)}")
        elif op == '%':
            print(f"Result: {num1 % num2}")
#handling operators that require one operand
    elif op in ['sqrt', 'log']:
        num = float(input("Enter number:"))
        if op == 'sqrt':
            if num >=0:
                print(f"Result:{math.sqrt(num)}")
            else:
                print("Enter valid number")
        if op == 'log':
            if num> 0:
                print(f"Result: {math.log10(num)}")
            else:
                print("Enter valid value")
    else:
        print("invalid operator")
mathe == False
```
